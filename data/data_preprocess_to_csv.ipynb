```python
path_name_test = '/home/ja768/Downloads/training data/SubLoc/test.pkl'
path_name_train = '/home/ja768/Downloads/training data/SubLoc/train.pkl'
path_name_valid = '/home/ja768/Downloads/training data/SubLoc/valid.pkl'
```

```python
import pickle

df_test = pd.read_pickle(path_name_test)
df_train = pd.read_pickle(path_name_train)
df_valid = pd.read_pickle(path_name_valid)
# # Load the pickle files
# with open(path_name_test, 'rb') as f:
#     df_test = pickle.load(f)

# with open(path_name_train, 'rb') as f:
#     df_train = pickle.load(f)

# with open(path_name_valid, 'rb') as f:
#     df_valid = pickle.load(f)

# Print the first 5 rows of each dataframe
print("Test DataFrame:")
print(df_test.head())
print("\nTrain DataFrame:")
print(df_train.head())
print("\nValid DataFrame:")
print(df_valid.head())

print(df_test[df_test['Location'] == 'Cell.membrane'].head(1))
```

```python
# Add split column to each dataframe
df_test['split'] = 'test'
df_train['split'] = 'train'
df_valid['split'] = 'valid'

# Combine all three dataframes
df_combined = pd.concat([df_test, df_train, df_valid], ignore_index=True)

# Select only sequence, label, split columns
df_output = df_combined[['Sequence', 'Location', 'split']].copy()
df_output.rename(columns={'Sequence': 'sequence', 'Location': 'label'}, inplace=True)

print("Combined dataframe shape:", df_output.shape)
print(df_output.head())

# Save to CSV
output_path = '/home/ja768/sc/peft_plm/data/raw/subcellular_localization.csv'
df_output.to_csv(output_path, index=False)
print(f"\nSaved to: {output_path}")
```
